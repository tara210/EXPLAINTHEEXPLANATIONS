```python
import pandas as pd
import spacy

nlp = spacy.load("en_core_web_sm")


def clean_tokens(tok):

    if tok.is_punct:
        return False

    return True


def lemmatize(tok):

    return tok.lemma_


# NOTE: fixed a bug here — this loop was previously indented inside
# lemmatize(), after its return statement, so it never ran. Also writes
# to a new results dict first rather than overwriting df["explanation_1"]
# in place while iterating over it, which would clobber values mid-loop.

column = "explanation_1"
results = {}

for index, value in df[column].items():

    text = []

    if pd.notna(value):

        for i in nlp(value):

            if clean_tokens(i):

                d = []

                d.append(i.text)
                d.append(lemmatize(i))
                d.append(i.pos_)

                text.append(d)

    results[index] = text

df[column] = pd.Series(results)

df.to_csv("eli5_comments_processed_clean.csv", index=False)
```

```python
import pandas as pd
import ast


# ============================================================
# 1. Load processed CSV
# ============================================================

df = pd.read_csv("eli5_comments_processed_clean.csv")


# ============================================================
# 2. Only explanation_1
# ============================================================

explanation_columns = ["explanation_1"]


# ============================================================
# 3. Targets
# ============================================================

phrase_targets = [
    "for example",
    "for instance",
    "e.g.",
    "i.e."
]

word_targets = [
    "imagine",
    "you",
    "illustrate",
    "demonstrate",
    "indicate"
]

verb_targets = [
    "mean",
    "show",
    "cause"
]

all_targets = (
    phrase_targets +
    word_targets +
    verb_targets
)


# ============================================================
# 4. Function for detecting token sequences
# ============================================================

def contains_phrase(words, phrase):

    phrase_tokens = phrase.lower().split()

    for i in range(
        len(words) - len(phrase_tokens) + 1
    ):

        if words[
            i:i + len(phrase_tokens)
        ] == phrase_tokens:

            return True

    return False


# ============================================================
# 5. Frequency dictionary
# ============================================================

frequency = {
    target: 0
    for target in all_targets
}


# ============================================================
# 6. Analyse each sample
# ============================================================

for index, row in df.iterrows():

    sample_tokens = []

    # --------------------------------------------------------
    # Only explanation_1
    # --------------------------------------------------------

    value = row["explanation_1"]

    if pd.isna(value):
        continue

    if value == "[]":
        continue

    try:

        tokens = ast.literal_eval(value)

        if isinstance(tokens, list):
            sample_tokens.extend(tokens)

    except (ValueError, SyntaxError):
        continue


    # --------------------------------------------------------
    # Get word sequence
    # --------------------------------------------------------

    words_in_sample = [
        str(token[0]).lower()
        for token in sample_tokens
        if len(token) >= 3
    ]


    # --------------------------------------------------------
    # Keep track of what was found in THIS sample
    # --------------------------------------------------------

    found_in_sample = set()


    # ========================================================
    # 7. Search multi-word phrases
    # ========================================================

    if contains_phrase(
        words_in_sample,
        "for example"
    ):
        found_in_sample.add("for example")


    if contains_phrase(
        words_in_sample,
        "for instance"
    ):
        found_in_sample.add("for instance")


    if contains_phrase(
        words_in_sample,
        "e.g."
    ):
        found_in_sample.add("e.g.")


    if contains_phrase(
        words_in_sample,
        "i.e."
    ):
        found_in_sample.add("i.e.")


    # ========================================================
    # 8. Search individual words
    # ========================================================

    for token in sample_tokens:

        if len(token) < 3:
            continue

        word = str(token[0]).lower()
        lemma = str(token[1]).lower()
        pos = str(token[2])


        # ----------------------------------------------------
        # "you"
        # ----------------------------------------------------

        if word == "you":
            found_in_sample.add("you")


        # ----------------------------------------------------
        # "imagine"
        # ----------------------------------------------------

        if lemma == "imagine":
            found_in_sample.add("imagine")


        # ----------------------------------------------------
        # illustrate / demonstrate / indicate
        # ----------------------------------------------------

        if lemma in [
            "illustrate",
            "demonstrate",
            "indicate"
        ]:
            found_in_sample.add(lemma)


        # ----------------------------------------------------
        # mean / show / cause
        #
        # ONLY if POS = VERB
        # ----------------------------------------------------

        if pos == "VERB":

            if lemma in [
                "mean",
                "show",
                "cause"
            ]:
                found_in_sample.add(lemma)


    # ========================================================
    # 9. Count each target ONCE per sample
    # ========================================================

    for target in found_in_sample:

        frequency[target] += 1


# ============================================================
# 10. Create frequency table
# ============================================================

frequency_df = pd.DataFrame(
    [
        {
            "target": target,
            "sample_frequency": frequency[target],
            "percentage": (
                frequency[target] / len(df)
            ) * 100
        }

        for target in all_targets
    ]
)


# ============================================================
# 11. Display
# ============================================================

print(frequency_df)


# ============================================================
# 12. Save
# ============================================================

frequency_df.to_csv(
    "explanation_1_lexical_analysis.csv",
    index=False
)

print(
    "\nSaved as: explanation_1_lexical_analysis.csv"
)
```
